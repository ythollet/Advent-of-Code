```python
def func_process_row(row: list[str]):

    # Conversion en int
    row = sorted([int(e) for e in row], reverse=True)

    for idx_n1, n1 in enumerate(row):
        for n2 in row[idx_n1+1:]:
            if n1 % n2 == 0:
                return n1 / n2
```

# Lecture et traitement du fichier source

```python
with open('real.txt') as f:
    file = sum(
        func_process_row(row.split())
        for row in f.read().splitlines()
    )

file
```
