```python
def func_get_ecart_row(row: list[str]):

    # Conversion en int
    row = [int(e) for e in row]

    # Détermine le min
    ecart_min_max = max(row) - min(row)

    return ecart_min_max
```

# Lecture et traitement du fichier source

```python
with open('real.txt') as f:
    file = sum(
        func_get_ecart_row(row.split())
        for row in f.read().splitlines()
    )

file
```
