# Lecture du fichier source

```python
with open('real.txt') as f:
    list_numbers = [int(digit) for digit in f.read()]

list_numbers
```

# Traitement

```python
ecart = int(len(list_numbers) / 2)

sum(
    current_number
    for current_number, next_number 
    in zip(list_numbers, list_numbers[ecart:]+list_numbers[0:ecart])
    if current_number == next_number
)
```
